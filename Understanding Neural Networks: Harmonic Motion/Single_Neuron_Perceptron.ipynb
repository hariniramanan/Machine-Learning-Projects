#Single Neuron Perceptron
In biology, we call the nueron the smallest functional unit of the nervous system, while the smallest unit of a neural network is a Perceptron, a single layer nueron. This notebook is an implementation using TensorFlow to understand how a basic neural network works.

The neuron calculates a weighted sum of the inputs, adds the bias, and runs it through the activation function:

$$z = (x_1 \cdot w_1) + (x_2 \cdot w_2) + \dots + b$$

$$\text{Output} = f(z) = \begin{cases} 1 & \text{if } z \geq 0 \\ 0 & \text{if } z < 0 \end{cases}$$

where w=weights and b=bias and x are inputs

Every time data passes through a neuron, it follows a strict timeline called Forward Propagation:

  1. The neuron accepts numerical features from your dataset ($x_1, x_2$).
  2. Each input is multiplied by its corresponding weight ($x \cdot w$). This scales the input based on its importance.
  3. The neuron adds all the weighted inputs together and adds the bias ($b$) to get the total net input ($z$).
  4. The net input $z$ passes through the step function. If $z$ is positive or zero, the neuron fires (1). If it's negative, it stays silent (0).


##Implementation
We will train our neuron to solve the AND gate problem we just walked through. Remember, an AND gate only outputs 1 if both inputs are 1

```python
import numpy as np

class Perceptron:
  def __init__(self, inputsize, learningrate=0.1):
    #Initialize weights randomly, bias=0
    self.weights = np.random.randn(inputsize)*0.01  #0.01 is a scaling factor to prevent 'exploding' inputs
    self.bias=0.0
    self.lr=learningrate

  def activation_function(self, z):
        # Step function, where output is 1 if function z>=0, else 0
        return 1 if z >= 0 else 0


  def predict(self, x):
        # Calculate z = (x1*w1 + x2*w2 + ...) + b
        z = np.dot(x, self.weights) + self.bias
        return self.activation_function(z)

  def train(self, training_inputs, labels, epochs=20):
        for epoch in range(epochs):
            total_error = 0 #loss
            for x, y in zip(training_inputs, labels):
                # 1. Forward Pass: Get prediction
                prediction = self.predict(x)

                # 2. Calculate Error
                error = y - prediction
                total_error += abs(error)

                # 3. Update Rule if there is an error
                if error != 0:
                    self.weights += self.lr * error * x
                    self.bias += self.lr * error

            # If the neuron makes 0 mistakes, we can stop training early!
            if total_error == 0:
                print(f"Accuracy reached at {epoch+1}!")
                break
```

##Testing the Perceptron on an AND Gate

```python
# 1. Dataset (Inputs: [x1, x2])
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# 2. True targets (y) for an AND gate
y = np.array([0, 0, 0, 1])

# 3. Create the neuron (2 inputs because our data has x1 and x2)
neuron = Perceptron(inputsize=2, learningrate=0.1)

print("Before training prediction for [1, 1]:", neuron.predict(np.array([1, 1])))

# 4. Train it!
neuron.train(X, y, epochs=20)

# 5. Verify the results
print("\nFinal Trained Predictions:")
for inputs in X:
    print(f"Input: {inputs} -> Model Predicted: {neuron.predict(inputs)}")
```
