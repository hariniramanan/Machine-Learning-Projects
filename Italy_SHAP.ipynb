```python
def avgKS(df1, df2):
  ksstats = []
  cols = (df1.columns)
  for col in cols:
    cdf1 = (df1[col].sort_values().cumsum())/(len(df1))
    cdf2 = (df2[col].sort_values().cumsum())/(len(df2))
    ksstats.append(np.max(np.abs(cdf1 - cdf2)))
  return np.average(ksstats)
```

```python
def computemetrics(y_true, y_proba):
    thresh = [0.15, 0.5]
    prop = np.mean(y_true == 1)  # prevalence
    auc = roc_auc_score(y_true, y_proba)
    for th in thresh:
        y_pred = (y_proba >= th).astype(int)
        tpr = recall_score(y_true, y_pred)  # sensitivity
        tnr = recall_score(y_true, y_pred, pos_label=0)  # specificity
        std_net_benefit = ((tpr * prop - (1 - tnr) * (1 - prop) * th / (1 - th)) / prop)
        bal_acc = balanced_accuracy_score(y_true, y_pred)
        print(
            f"Threshold - {th} Standardized Net Benefit - {round(std_net_benefit, 3)}",
            f"AUC Score - {round(auc, 3)}",
            f"Balanced Accuracy - {round(bal_acc, 3)}"
        )

computemetrics(Y_test, y_proba)
```

```python
# ============ SHAP CELL 1 : imports and helpers ============
import os
import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

OUT = '../SHAP_output'
os.makedirs(OUT, exist_ok = True)

def shap_for(model, X):
    """Return SHAP values and the frame they were computed on."""
    if hasattr(model, 'named_steps'):                  # India models are Pipelines
        X = X[list(map(str, model.feature_names_in_))]
        sel = model.named_steps['select']
        est = model.named_steps['model']
        cols = [c for c, k in zip(X.columns, sel.get_support()) if k]
        Xe = pd.DataFrame(sel.transform(X), columns = cols, index = X.index)
    else:                                              # Italy / Brazil are plain models
        est = model
        Xe = X[list(map(str, model.feature_names_in_))]
    sv = np.array(shap.TreeExplainer(est).shap_values(Xe))
    if sv.ndim == 3:
        sv = sv[:, :, 1]
    return sv, Xe

def report(tag, model, X):
    """Print the feature weight table and save CSV + 300 dpi figures."""
    sv, Xe = shap_for(model, X)
    t = pd.DataFrame({'feature': Xe.columns, 'mean_abs_shap': np.abs(sv).mean(0)})
    t = t.sort_values('mean_abs_shap', ascending = False).reset_index(drop = True)
    t['share_%'] = (t['mean_abs_shap'] / t['mean_abs_shap'].sum() * 100).round(1)
    print('\\n' + tag + '  (n = ' + str(len(Xe)) + ', ' + str(Xe.shape[1]) + ' features)')
    print(t.to_string(index = False))
    t.to_csv(OUT + '/shap_' + tag + '.csv', index = False)

    plt.figure()
    shap.summary_plot(sv, Xe, plot_type = 'bar', show = False, max_display = 14)
    plt.tight_layout(); plt.savefig(OUT + '/bar_' + tag + '.png', dpi = 300); plt.show()

    plt.figure()
    shap.summary_plot(sv, Xe, show = False, max_display = 14)
    plt.tight_layout(); plt.savefig(OUT + '/beeswarm_' + tag + '.png', dpi = 300); plt.show()
    return t

print('shap', shap.__version__, '| helpers ready')
```

```python
# ============ SHAP CELL 2 : rebuild the Indian 375 test frame ============
# The other three test cohorts are still in memory from this notebook:
#   dfItaly  -> Italian cohort          (cell 8 to 14)
#   df2      -> Indian SMOTE 600        (cell 29, already z-scored)
#   dfBrazil -> Brazilian cohort        (cell 37 to 40)
# Only the Indian 375 frame was overwritten, so it is rebuilt here.
# Same steps as cells 4 to 7 and cell 15, unchanged.

df1_shap = pd.read_csv('../Data/Edited_bitsM.csv')
df2_shap = pd.read_csv('../Data/ESIdf.csv')
df1_shap = df1_shap[df1_shap['Result'] != 'Positive']       # only the 300 negatives
df_shap = pd.concat([df1_shap, df2_shap], join = 'inner')
df_shap = df_shap.drop(columns = [df_shap.columns[0], 'Others'])

encoder_shap = LabelEncoder()
df_shap['Result'] = encoder_shap.fit_transform(df_shap['Result'])
df_shap['Gender'] = encoder_shap.fit_transform(df_shap['Gender'])

df_shap = df_shap[order].copy()          # `order` is still defined, from cell 15
cols_shap = df_shap.columns.tolist()
cols_shap.remove('Result')

# Standard Scaling of Indian dataset  (identical to cell 15)
for c in cols_shap:
    df_shap[c] = ((df_shap[c] - (df_shap[c].mean())))/(df_shap[c].std())

X_ind375 = df_shap.drop(columns = ['Result'])
print('Indian 375 frame rebuilt :', X_ind375.shape)
print('Indian 600 (df2, in memory):', df2.drop(columns = ['Result']).shape)
print('Italian   (dfItaly)        :', dfItaly.drop(columns = ['Result']).shape)
print('Brazilian (dfBrazil)       :', dfBrazil.drop(columns = ['Result']).shape)
```

```python
# ============ SHAP CELL 3 : run SHAP on every cohort this model is tested on ============
# `model` is the Italy model, loaded in cell 20.
t_ita  = report('Italy_on_Italy',     model, dfItaly.drop(columns = ['Result']))
t_i375 = report('Italy_on_India_375', model, X_ind375)
t_i600 = report('Italy_on_India_600', model, df2.drop(columns = ['Result']))
t_bra  = report('Italy_on_Brazil',    model, dfBrazil.drop(columns = ['Result']))
```

```python
# ============ SHAP CELL 4 : comparison across cohorts ============
comp = pd.DataFrame({
    'Italy_%':     t_ita.set_index('feature')['share_%'],
    'India375_%':  t_i375.set_index('feature')['share_%'],
    'India600_%':  t_i600.set_index('feature')['share_%'],
    'Brazil_%':    t_bra.set_index('feature')['share_%'],
})
comp = comp.sort_values('Italy_%', ascending = False)
comp.to_csv(OUT + '/italy_model_all_cohorts.csv')
print('ITALY MODEL: share of total |SHAP| on each test cohort')
print(comp.to_string())
print()

# the transfer the professor asked about: Italian training data vs Indian test data
shift = comp[['Italy_%', 'India375_%']].copy()
shift['change'] = shift['India375_%'] - shift['Italy_%']
shift = shift.sort_values('change', ascending = False)
shift.to_csv(OUT + '/italy_transfer_shift.csv')
print('ITALY MODEL: Italian input vs Indian input')
print(shift.to_string())

ax = shift['change'].plot(kind = 'barh', figsize = (7, 5))
ax.axvline(0, color = 'black', lw = 0.8)
ax.set_xlabel('change in share of total |SHAP| (percentage points)')
ax.set_title('Italy model: Italian input vs Indian input')
plt.tight_layout(); plt.savefig(OUT + '/italy_transfer_shift.png', dpi = 300); plt.show()
```
